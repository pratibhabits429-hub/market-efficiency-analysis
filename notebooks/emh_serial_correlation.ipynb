# Market Efficiency Analysis: Serial Correlation Test

## Overview
This notebook tests weak-form market efficiency by examining whether daily Nifty 50 returns show statistically significant serial dependence.

The analysis uses lag-1 autocorrelation and the Ljung–Box test. Under weak-form efficiency, past returns should not provide statistically significant information about future returns.

**Data source:** Yahoo Finance via the `yfinance` Python package.
**Market:** Nifty 50 (`^NSEI`)
**Frequency:** Daily
**Sample:** Five years of available observations at the time the notebook is run.

> Because the notebook downloads live market data, numerical results can change slightly when it is rerun.

## 1. Install and import packages

Run the installation cell first if `yfinance` is not already available.

```python
!pip install yfinance statsmodels --quiet

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.stats.diagnostic import acorr_ljungbox
```

## 2. Download market data

```python
data = yf.download("^NSEI", period="5y", interval="1d", auto_adjust=False)

data.head()
```

## 3. Calculate daily returns

Returns are calculated as the percentage change in adjusted closing prices.

```python
data['Return'] = data['Adj Close'].pct_change() * 100
returns = data['Return'].dropna()

returns.describe()
```

## 4. Lag-1 autocorrelation

The first-order autocorrelation measures the relationship between today's return and the previous day's return.

A value close to zero is consistent with little linear serial dependence.

```python
lag1_acf = returns.autocorr(lag=1)
print("Lag-1 autocorrelation:", lag1_acf)
```

## 5. Autocorrelation function (ACF)

```python
plot_acf(returns, lags=20)
plt.title("Autocorrelation of Nifty 50 Daily Returns")
plt.show()
```

## 6. Ljung–Box test

The Ljung–Box test examines whether a group of autocorrelations is jointly different from zero. Here, the first 10 lags are tested.

```python
ljung_box = acorr_ljungbox(returns, lags=[10], return_df=True)
print(ljung_box)
```

## 7. Interpretation

The null hypothesis of the Ljung–Box test is that the tested autocorrelations are jointly zero.

- A lag-1 autocorrelation close to zero indicates limited first-order serial dependence.
- A Ljung–Box p-value above 0.05 means there is insufficient evidence to reject the null hypothesis of no serial correlation over the tested lags.

Together, these results can be interpreted as evidence consistent with weak-form market efficiency, subject to the limitations of the sample and methodology.
